```python
from pymongo import MongoClient
from dotenv import load_dotenv
import pandas as pd
import os

load_dotenv(".env")

client = MongoClient(os.getenv("MONGO_URI"))

db = client[os.getenv("MONGO_DB_NAME")]

coleccion = db[os.getenv("MONGO_COLLECTION_PROCESSED")]

datos = list(
    coleccion.find(
        {},
        {
            "_id": 0,
            "precio": 1,
            "m2": 1,
            "dormitorios": 1,
            "banos": 1
        }
    )
)

df = pd.DataFrame(datos)

print(df.shape)

df.head()
```

```python
df = df.dropna()

df = df[
    (df["precio"] > 0)
    & (df["m2"] > 0)
]

print(df.shape)
```

```python
correlacion = df[
    ["precio", "m2", "dormitorios", "banos"]
].corr(method="pearson")

correlacion
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

plt.figure(figsize=(8,6))

sns.heatmap(
    correlacion,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Matriz de Correlación de Pearson")

plt.show()
```

```python
df["precio_m2"] = df["precio"] / df["m2"]

df[["precio","m2","precio_m2"]].head()
```

```python
mini = df[df["m2"] < 40]

grandes = df[df["m2"] > 80]

print("Mini departamentos:", len(mini))
print("Grandes propiedades:", len(grandes))
```

```python
resultado = pd.DataFrame({
    "Grupo": ["Menos de 40 m²", "Más de 80 m²"],
    "Precio Promedio por m²": [
        mini["precio_m2"].mean(),
        grandes["precio_m2"].mean()
    ]
})

resultado
```

```python
plt.figure(figsize=(8,5))

sns.barplot(
    data=resultado,
    x="Grupo",
    y="Precio Promedio por m²"
)

plt.title("Comparación de Precio por m²")

plt.ylabel("Precio por m²")

plt.show()
```
